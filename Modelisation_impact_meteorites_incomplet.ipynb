### Suggestion : utiliser la fonction logarithme décimal pour la modélisation

Si la loi de puissance s’écrit : $D = a\, E^b$

alors, on peut écrire : $\log(D) = \log(a) + b\, \log(E)$

On obtient ainsi une relation linéaire entre : $y = \log(D)$ et $x = \log(E)$.

La régression linéaire de $y$ en fonction de $x$, permet alors de déterminer : 
- la pente de la droite, égale à $b$ ;
- l’ordonnée à l’origine $\log(a)$.

#### Ajustement par régression linéaire - Utilisation de la fonction log

En prenant le logarithme de la loi de puissance $D = a,E^b$, on obtient une relation linéaire $\log(D) = \log(a) + b,\log(E)$ dont la régression permet de déterminer l’exposant $b$ (pente) et la constante $a$ (ordonnée à l’origine).

```python
# Ajustement par régression linéaire - Utilisation de la fonction log

# Linéarisation : log10(D) = log10(a) + b*log10(Ei_mJ)
X = np.log10(Ei_mJ)
Y = np.log10(D_cm)

# Régression linéaire : Y = alpha + b*X
b, alpha = np.polyfit(X, Y, 1)

# Constante du modèle
a = 10**alpha

print(f"Paramètres ajustés : a = {a:.3g}, b = {b:.4f}")

# ===== Tracé =====
plt.figure()
plt.scatter(X, Y, color="blue", marker="o", label="Données expérimentales")
plt.plot(X, alpha + b*X, color="purple", linestyle="-", label="Modélisation")
plt.xlabel("$log(E_i),\ Ei\ en\ mJ$")
plt.ylabel("$log(D),\ D\ en\ cm$")
plt.title("Modélisation par une fonction linéaire des mesures du diamètre du cratère - Fonction log")
plt.grid(True)
plt.legend()
plt.show()
```

### Critère de validité (statistique) : moyenne, écart-type et écart-normalisé

Dans la cellule suivante, renseigner les valeurs de l’exposant $b$ obtenues par chaque binôme dans la liste Python dédiée. *Cette liste sera ensuite transformée en tableau afin de faciliter les calculs statistiques.*

À partir de l’ensemble des valeurs de $b$, le programme permettra de calculer :
- la **valeur moyenne $\bar b$**, assimilée au résultat de la mesure $b_{mes}$ ;
- l’**écart-type expérimental $\sigma_{n-1}$**, qui caractérise la dispersion des résultats ;
- l’**incertitude-type** sur la moyenne $u(\bar b)=\dfrac{\sigma_{n-1}}{\sqrt{n}}$, *avec $n$ le nombre de mesures dans l'échatillon*.

Enfin, on calculera le **z-score** (ou écart-normalisé) permettant de tester la **compatibilité du résultat de la mesure avec la valeur attendue $b_{\text{att}}=0{,}25$** :
$$z = \dfrac{|b_{mes} - b_{\text{att}}|}{u(\bar{b})}$$
Selon le critère usuel, si $z\leq2$, le résultat de la mesure est compatible avec la valeur attendue, la modélisation est donc acceptable.


```python
# Liste des valeurs de b obtenues par chaque binôme - Chaque binôme complète ou modifie cette liste avec sa valeur mesurée
b_liste = [..., ..., ..., ..., ..., ..., ..., ...] # (à compléter)

# Conversion de la liste en tableau NumPy pour faciliter les calculs
b_vals = np.array(b_liste)

# Nombre de mesures
n = len(b_vals)

# Calcul de la valeur moyenne (résultat de la mesure)
b_mean = b_vals.mean()

# Calcul de l'écart-type expérimental (dispersion des mesures)
# ddof=1 correspond à l'écart-type de l'échantillon : s_(n-1)
s_n1 = b_vals.std(ddof=1)

# Calcul de l'incertitude-type sur la moyenne
u_mean = s_n1 / np.sqrt(n)

# Valeur attendue de l'exposant b (issue du modèle théorique)
b_att = 0.25

# Calcul du z-score pour tester la compatibilité
z = abs(b_mean - b_att) / u_mean

# Affichage des résultats
print(f"Nombre de mesures n = {n}")
print(f"Moyenne b̄ = {b_mean:.5f}")
print(f"Écart-type s_(n-1) = {s_n1:.5f}")
print(f"Incertitude-type u(b̄) = {u_mean:.5f}")
print(f"z-score (comparaison à b_att = 0.25) : z = {z:.3f}")

# Test de compatibilité (critère usuel)
if z <= 2:
    print("Conclusion : la valeur mesurée de b est compatible avec la valeur attendue !")
else:
    print("Conclusion : la valeur mesurée de b n'est pas compatible avec la valeur attendue.")
```
